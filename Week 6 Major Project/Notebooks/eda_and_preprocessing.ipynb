```python
# Importing Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import warnings
warnings.filterwarnings("ignore")
from sklearn.preprocessing import StandardScaler
```

```python
# Loading dataset
df = pd.read_csv("../Datasets/creditcard.csv")
```

```python
# Class Distribution Plot
plt.figure(figsize=(8,6))
sns.countplot(x="Class", data=df)
plt.xticks(ticks=[0, 1], labels=["No", "Yes"])
plt.title("Class Distribution (0: Not Fraud, 1: Fraud)")
plt.tight_layout()
plt.show()

print(df["Class"].value_counts(normalize=True))
```

```python
# Correlation Matrix
plt.figure(figsize=(12,10))
sns.heatmap(df.corr(), annot=False, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

```python
# Boxplots
plt.figure(figsize=(8, 6))
plt.subplot(3, 2, 1)
sns.boxplot(x='Class', y='Amount', data=df)
plt.title('Amount by Class')
plt.xlabel('Class')
plt.ylabel('Amount')

plt.subplot(3, 2, 2)
sns.boxplot(x='Class', y='Time', data=df)
plt.title('Time by Class')
plt.xlabel('Class')
plt.ylabel('Time')

plt.show()
```

```python
# PCA Visualization for n = 2 components
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
principalComponents = pca.fit_transform(df.drop('Class', axis=1))
plt.figure(figsize=(8,6))
plt.scatter(principalComponents[:, 0], principalComponents[:, 1], c=df['Class'], cmap='coolwarm', alpha=0.5)
plt.title('PCA of Transactions')
plt.show()
```

```python
# Normalizing the Amount and Time features
df['Amount'] = StandardScaler().fit_transform(df['Amount'].values.reshape(-1,1))
df['Time'] = StandardScaler().fit_transform(df['Time'].values.reshape(-1,1))
#df = df.drop(['Amount', 'Time'], axis=1)
```
