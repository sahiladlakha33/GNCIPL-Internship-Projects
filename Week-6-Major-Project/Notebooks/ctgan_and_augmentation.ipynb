```python
# Importing libraries
from ctgan import CTGAN
import pandas as pd
```

```python
# Loading real dataset after preprocessing
df = pd.read_csv("../Datasets/creditcard_preprocessed.csv")
```

```python
# Using GenAI Model for Oversampling of Class = 1(fraud)
ctgan = CTGAN(epochs=50, batch_size=100, verbose=True)
ctgan.fit(df, discrete_columns=['Class'])
```

```python
synthetic_data = ctgan.sample(5000)
```

```python
synthetic_data['Class'] = 1
```

```python
synthetic_data.head()
```

```python
# Saving the synthetic data
synthetic_data.to_csv('../Datasets/creditcard_synthetic.csv', index=False)
```

```python
# Combine original and synthetic data
augmented_df = pd.concat([df, synthetic_data], ignore_index=True)

# Shuffle the dataset for randomness
augmented_df = augmented_df.sample(frac=1, random_state=42).reset_index(drop=True)

# Save augmented dataset
augmented_df.to_csv('../Datasets/creditcard_augmented.csv', index=False)
```
